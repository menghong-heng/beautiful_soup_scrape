# Web Scrapping & Combine all CSV together

## Import Necessary Libraries

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import glob
import os
```

### 1 Star of Amazon (101 pages)

```python
base_url = "https://www.trustpilot.com/review/www.amazon.com?stars=1&page="
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; WOW64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/137.0.0.0 Safari/537.36'}

all_reviews = []
max_pages = 105  # increase this to get more reviews

for page in range(2, max_pages + 1):
    url = base_url + str(page)
    print(f"Scraping page {page}...")

    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, 'html.parser')

    cards = soup.find_all('article')

    if not cards:
        print("No more reviews found.")
        break

    for card in cards:
        # Review text
        text_tag = card.find('p', {'data-service-review-text-typography': 'true'})
        text = text_tag.get_text(strip=True) if text_tag else None

        # Star rating
        star_img = card.find('img', alt=True)
        stars = None
        if star_img:
            alt_text = star_img['alt']
            if "Rated" in alt_text:
                try:
                    stars = int(alt_text.split(' ')[1])
                except:
                    stars = None

        if text and stars:
            all_reviews.append({'text': text, 'stars': stars})

    time.sleep(1)  # Be respectful to Trustpilot

# Create DataFrame
df = pd.DataFrame(all_reviews)
df['sentiment'] = df['stars'].apply(lambda x: 'positive' if x >= 4 else 'neutral' if x == 3 else 'negative')

df.to_csv('trustpilot_amazon_1_star_reviews.csv', index=False)
print(f"Done! Scraped {len(df)} reviews.")
print(df.head())
```

## Combine all CSV files together

```python
# Get all CSVs matching the pattern
files = glob.glob("trustpilot_*_star_reviews.csv")

# Load and label each file
all_data = []
for file in files:
    df = pd.read_csv(file)

    # Add sentiment column if missing
    if 'sentiment' not in df.columns:
        df['sentiment'] = df['stars'].apply(
            lambda x: 'negative' if x <= 2 else 'neutral' if x == 3 else 'positive'
        )

    # Add source file as origin (optional)
    df['source_file'] = os.path.basename(file)

    all_data.append(df)

# Combine all into one DataFrame
combined_df = pd.concat(all_data, ignore_index=True)

# Remove duplicates based on review text
combined_df.drop_duplicates(subset='text', inplace=True)

# Save to file
combined_df.to_csv("trustpilot_all_sources_combined.csv", index=False)

print(f"Merged {len(files)} files into {len(combined_df)} unique reviews.")
```
